```python
corr = materials.corr(numeric_only=True)

# unstack turns the grid into a flat list of pairs, then we sort by strength
corr_pairs = corr.unstack().sort_values(key=abs, ascending=False)

# drop self-correlations (always 1.0) and duplicate pairs (A-B same as B-A)
corr_pairs = corr_pairs[corr_pairs != 1.0]
corr_pairs = corr_pairs[~corr_pairs.index.duplicated()]

corr_pairs.head(15)
```

## Findings

**Missing values:** One incomplete row found in materials.csv (MAT_0404, 
14/23 values missing) — dropped. No missing values in products.csv.

**Duplicates:** None found in either dataset (checked on material_id / product_id).

**Outliers:**
- biodegradation_time_days ranges from 31 to 180,208 days. This is driven 
  entirely by material_type — Cardboard/Paper are in the hundreds of days, 
  Plastic/Steel are in the tens of thousands. Not an error, but will need 
  a log transform before modeling.
- cost_per_unit_usd is right-skewed, with Steel pulling the tail up to ~$26.

**Data quality concerns:**
- Inconsistent category values in products.csv (" Cosmetics" vs "Cosmetics") — fixed.
- Column names in materials.csv used inconsistent casing/symbols — standardized to snake_case.
- Class imbalance in material_type: Plastic (199), Cardboard (139), 
  Paper/Bio-Based (37), Steel (29) — predictions for underrepresented 
  types may be less reliable.

**Correlations:** Several features are highly correlated with each other 
(e.g. moisture/thermal resistance, co2_emission_per_kg with several others), 
suggesting multicollinearity. Not a blocker for tree-based models 
(Random Forest/XGBoost) but worth noting.

**Next step:** Proceed to feature engineering — log-transform 
biodegradation_time_days, encode categorical variables, and build the 
CO2 Impact Index / Cost Efficiency Index mentioned in the project brief.
